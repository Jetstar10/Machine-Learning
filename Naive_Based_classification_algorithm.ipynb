```python
import pandas as pd
df= pd.read_csv('data/titanic2.csv')
df.head()
```

```python
df.drop(['PassengerId', 'Name', 'SibSp', 'Parch', 'Ticket', 'Cabin', 'Embarked'], axis='columns', inplace=True)
df.head()
```

```python
target = df.Survived
target
```

```python
inputs = df.drop(['Survived'], axis='columns')
inputs
```

```python
dummies = pd.get_dummies(inputs.Sex)
bool_cols = dummies.select_dtypes(include="bool").columns
dummies[bool_cols] = dummies[bool_cols].astype(int)
dummies
```

```python
inputs = pd.concat([inputs, dummies], axis='columns')
inputs.head()
```

```python
inputs.drop('Sex', axis='columns', inplace=True)
inputs
```

```python
inputs.columns[inputs.isna().any()]
```

```python
inputs.Age[:10]
```

```python
mean_age = round(inputs.Age.mean())
inputs.Age = inputs.Age.fillna(mean_age)
inputs.Age[:10]
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(inputs,target, test_size=0.2)
len(x_train)
```

```python
len(x_test)
```

```python
len(inputs)
```

```python
from sklearn.naive_bayes import GaussianNB
model = GaussianNB()
model.fit(x_train,y_train)
```

```python
model.score(x_test,y_test)
```

```python
x_test[:10]
```

```python
y_test[:10]
```

```python
model.predict(x_test[:10])
```

```python
model.predict_proba(x_test[:10])
```

```python
import pandas as pd
df = pd.read_csv('data/spam.csv')
df.head()
```

```python
df.groupby('Category').describe()
```

```python
df['spam'] = df['Category'].apply(lambda x:1 if x == 'spam' else 0)
df.drop(['Category'], axis='columns', inplace=True)
df
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(df.Message, df.spam, test_size=0.25)
x_train
```

```python
from sklearn.feature_extraction.text import CountVectorizer
c = CountVectorizer()
x_train_count = c.fit_transform(x_train.values)
x_train_count.toarray()[:3]
```

```python
from sklearn.naive_bayes import MultinomialNB
model = MultinomialNB()
model.fit(x_train_count,y_train)
```

```python
emails = [
    'Hey mohan, can we get together to watch football tomorrow?',
    'Up to 20% discount on parking , exclusive offer just for you. Dont miss this reward!'
]
emails_count = c.transform(emails)
model.predict(emails_count)
```

```python
x_train_count = c.transform(x_test)
model.score(x_train_count, y_test)
```

```python
from sklearn.pipeline import Pipeline
clf = Pipeline([
    ('vectorizer', CountVectorizer()),
    ('nb', MultinomialNB())
])
```

```python
clf.fit(x_train, y_train)
```

```python
clf.score(x_test, y_test)
```

```python
clf.predict(emails)
```
