```python
import pandas as pd
df = pd.read_csv('data/homeprices2.csv')
print(df)
```

```python
dummies = pd.get_dummies(df.town)
dummies
```

```python
merge = pd.concat([df, dummies],axis='columns')
print(merge)
```

```python
final = merge.drop(['town','west windsor'], axis='columns')
final
```

```python
from sklearn import linear_model
reg = linear_model.LinearRegression()
```

```python
x = final.drop('price', axis='columns')
```

```python
y = final.price
```

```python
reg.fit(x,y)
```

```python
reg.predict([[2800,0,1]])
```

```python
reg.predict([[3400,0,0]])
```

```python
reg.score(x,y)
```

```python
df = pd.read_csv('data/carprices.csv')
df
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
dummies = pd.get_dummies(df['Car Model'])
dummies
```

```python
merged = pd.concat([df,dummies],axis='columns')
merged
```

```python
final = merged.drop(["Car Model","Mercedez Benz C class"],axis='columns')
final
```

```python
df2 = final
df2[df2.select_dtypes("bool").columns] = df2.select_dtypes("bool").astype(int)
df2
```

```python
X = final.drop('Sell Price($)',axis='columns')
X
```

```python
y = final['Sell Price($)']
y
```

```python
from sklearn import linear_model
reg = linear_model.LinearRegression()
reg.fit(X,y)
```

```python
reg.predict([[45000,4,0,0]])
```

```python
reg.predict([[86000,7,0,1]])
```

```python
reg.score(X,y)
```
