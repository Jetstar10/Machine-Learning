```python
# Coefficients are the weight or the gradient of the best-fit lines
# Features are the independent variables in the equation

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn import linear_model

df = pd.read_csv("data/MachineLearning.csv")

plt.xlabel("area(sqr ft)")
plt.ylabel("price($SGD)")
plt.scatter(df.area, df.price, color="red", marker="+")

reg = linear_model.LinearRegression()
reg.fit(df[['area']],df.price)
```

```python
reg.predict([[4000]])
```

```python
reg.coef_
```

```python
reg.intercept_
```

```python
plt.xlabel('area', fontsize=20)
plt.ylabel('area', fontsize=20)
plt.scatter(df.area,df.price,color='red',marker='+')
plt.plot(df.area,reg.predict(df[['area']]),color='blue')
```

```python
d = pd.read_csv("data/areas.csv")
d.head(3)
```

```python
p = reg.predict(d)
d['prices'] = p
d.to_csv('prediction.csv', index=False)
```

```python
df = pd.read_csv("data/homeprices.csv")
print(df)
```

```python
reg = linear_model.LinearRegression()

# To create a data frame using existing data frame, u use double square brackets. The ones outside the bracket is the target ur trying to predict.

reg.fit(df[['area','bedrooms','age']], df.price)
```

```python
df = pd.read_csv("data/exercise2.csv")
print(df)
```

```python
from word2number import w2n
df['experience']= df['experience'].apply(lambda x: w2n.word_to_num(x) if isinstance(x, str) else x)
df['experience'] = df['experience'].fillna(0)
test_scores = math.floor(df['test_score(out of 10)'].mean())
print(test_scores)
```

```python
df['test_score(out of 10)'] = df[['test_score(out of 10)']].fillna(test_scores)
print(df)
```

```python
reg = linear_model.LinearRegression()
reg.fit(df[['experience', 'test_score(out of 10)', 'interview_score(out of 10)']], df['salary($)'])
```

```python
reg.predict([[2,9,6]])
```

```python
reg.predict([[12,10,10]])
```
